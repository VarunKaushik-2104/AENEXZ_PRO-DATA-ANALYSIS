```python
def rev(a):
    if a == "":
        return ""
    else:
        return a[-1] + rev(a[:-1])
print(rev("Aenexz"))
```

```python
a,b,c = map(int, input("Enter 3 No's: ").split())
if a > b and a > c:
    print(a, "is the largest")
elif b > a and b > c:
    print(b, "is the largest")
else:
    print(c, "is the largest")
```

```python
n=int(input("Enter a Year: "))
if(n%4==0 and n%100!=0) or (n%400==0):
    print(n,"is a Leap Year")
else:
    print(n,"is not a Leap Year")
```

```python
n=input("Enter a character: ")
if n in ('a','e','i','o','u'):
    print(n,"is a vowel")
else:
    print(n,"is a consonant")
```

```python
n=int(input("Enter a number: "))
if n%5==0 and n%11==0:
    print(n,"is divisible by 5 and 11")
elif n%5==0:
    print(n,"is divisible by 5")
elif n%11==0:
    print(n,"is divisible by 11")
else:
    print(n,"is not divisible by 5 and 11")
```

```python
n=int(input("Enter a number: "))
num=n
sum=0
while n>0:
    sum+=n
    n-=1
print("Sum of first", num, "natural numbers is:", sum)
```

```python
n=int(input("Enter number of steps: "))
for i in range(1, n+1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
n=int(input("Enter number of steps: "))
for i in range(1, n+1):
    for j in range(1, i+1):
        print(j)
    print("\n")
```
