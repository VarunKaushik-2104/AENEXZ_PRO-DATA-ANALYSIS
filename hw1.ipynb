```python
n=int(input("Enter no of Units: "))
if n>0 and n<100:
    print(f"Electricity Bill: ₹{5*n}")
elif n>100 and n<200:
    print(f"Electricity Bill: ₹{7*n}")
elif n>200:
    print(f"Electricity Bill: ₹{10*n}")

```

```python
amt=int(input("Enter the Purchase amount: "))
if amt>=5000:
    discount=amt*0.2
elif amt>=2000 and amt<5000:
    discount=amt*0.1
else:
    discount=0
print(f"Original Amount: ₹{amt}")
print(f"Discount Amount: ₹{discount:.2f}")
print(f"Final Amount Payable: ₹{amt - discount:.2f}")
```

```python
age=int(input("Enter your age: "))
salary=int(input("Enter your salary: "))
credit_score=int(input("Enter your credit score: "))
if (age>=21 and age<=60) and salary>=30000:
    if credit_score>=750:
        print("Loan Approved")
    elif credit_score>=650 and credit_score<750:
        print("Loan Approved with Guarantor")
    else:
        print("Loan Rejected")
else:
    print("You are not eligible for the Loan.")
```
